```python
from scipy.sparse import save_npz
from scipy.sparse import csr_matrix
from collections import Counter


def str_para_categorico(df: pd.DataFrame, coluna: str, k: int) -> pd.DataFrame:
    """
    Converte atributo de string para matriz esparsa
    """
    c = Counter()
    for string in df[coluna]:
        if string is not np.nan:
            c.update(string.split(", "))
    res = pd.DataFrame(index=df["title"])

    n_cols = 0
    for i in c.most_common():
        n_cols += 1
        if i[1] < k:
            break
    print(coluna, n_cols, "colunas")

    for i in c.most_common():
        if i[1] < k:
            break
        nova_col = []
        for string in df[coluna]:
            if string is not np.nan:
                nova_col.append(1 if i[0] in string.split(", ") else 0)
            else:
                nova_col.append(0)
        res[i[0]] = np.array(nova_col, dtype=np.int8)
        res = res.copy()

    sparse = csr_matrix(res)
    save_npz(f"matrizes_esparsas/sparse_{coluna}_{k}.npz", sparse) # Salva em forma de matriz esparsa scipy

    return res
```

```python
from scipy.sparse import load_npz
from scipy.sparse import hstack
import os

def abrir_matrizes_esparsas(k: int = 5):
    """
    Carrega e concatena matrizes esparsas
    """
    l = []
    for arquivo in os.listdir("./matrizes_esparsas"):
        try:
            k_ok = int(arquivo.split("_")[-1][:-4]) == k
        except:
            k_ok = False
        if arquivo[-4:] == ".npz" and k_ok:
            l.append(load_npz(f"./matrizes_esparsas/{arquivo}"))
            print(arquivo)
    return hstack(l)
```

# Testes dos algoritmos de redução de dimensionalidade PCA, UAP e tse-n

```python
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
import numpy as np

def cluster_and_plot(X, n_clusters):
    """
    Aplica KMeans em um dataset e plota um gráfico de dispersão dos clusters.

    Parâmetros:
    - X: matriz de entrada (numpy array ou DataFrame) com n amostras e m atributos.
    - n_clusters: número de clusters desejado.
    """
    # Aplica o KMeans
    kmeans = KMeans(n_clusters=n_clusters, random_state=42)
    labels = kmeans.fit_predict(X)

    # Reduz para 2D se necessário (para visualização)
    if X.shape[1] > 2:
        pca = PCA(n_components=2)
        X_vis = pca.fit_transform(X)
        print("Redução de dimensionalidade com PCA aplicada para visualização.")
    else:
        X_vis = X

    # Plotando os clusters
    plt.figure(figsize=(8, 6))
    scatter = plt.scatter(X_vis[:, 0], X_vis[:, 1], c=labels, cmap='viridis', s=50)
    centers = kmeans.cluster_centers_

    # Se foi feita redução, também transformamos os centros
    if X.shape[1] > 2:
        centers = pca.transform(centers)

    plt.scatter(centers[:, 0], centers[:, 1], c='red', s=200, alpha=0.75, marker='X', label='Centros')
    plt.title(f'Clusterização com {n_clusters} clusters')
    plt.xlabel('Componente 1')
    plt.ylabel('Componente 2')
    plt.legend()
    plt.grid(True)
    plt.show()

cluster_and_plot(sparse, 5)
```

```python
# prompt: Crie uma função de machine learning para agrupamento usando o algoritmo Kmeans e o t-SNE para redução de dimensionalidade

from sklearn.manifold import TSNE
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
import numpy as np
import seaborn as sns # Optional: for potentially nicer plots

def cluster_and_plot_tsne(X, n_clusters, perplexity=30.0, n_iter=300):
    """
    Aplica redução de dimensionalidade com t-SNE, KMeans e plota o resultado.

    Parâmetros:
    - X: matriz de entrada (numpy array ou DataFrame) com n amostras e m atributos.
    - n_clusters: número de clusters desejado.
    - perplexity: parâmetro perplexity para t-SNE.
    - n_iter: número de iterações para t-SNE.
    """
    # Redução de dimensionalidade com t-SNE
    print(f"Aplicando t-SNE com perplexity={perplexity}, n_iter={n_iter}...")
    # Fix: Use init="random" for sparse input
    tsne = TSNE(n_components=2, random_state=42, perplexity=perplexity, n_iter=n_iter, init="random")
    X_tsne = tsne.fit_transform(X)
    print("t-SNE concluído.")

    # Aplica o KMeans nos dados reduzidos
    print(f"Aplicando KMeans com n_clusters={n_clusters}...")
    kmeans = KMeans(n_clusters=n_clusters, random_state=42, n_init=10)
    labels = kmeans.fit_predict(X_tsne)
    print("KMeans concluído.")

    # Plotando os clusters
    plt.figure(figsize=(10, 8))
    scatter = plt.scatter(X_tsne[:, 0], X_tsne[:, 1], c=labels, cmap='viridis', s=10)
    centers = kmeans.cluster_centers_

    plt.scatter(centers[:, 0], centers[:, 1], c='red', s=200, alpha=0.75, marker='X', label='Centros')
    plt.title(f'Clusterização com {n_clusters} clusters (t-SNE)')
    plt.xlabel('Componente t-SNE 1')
    plt.ylabel('Componente t-SNE 2')
    plt.legend()
    plt.grid(True)
    plt.colorbar(scatter, label='Cluster')
    plt.show()

# Exemplo de uso com t-SNE:
# Use a matriz esparsa 'sparse' criada no código anterior
n_clusters_escolhido = 5 # Substitua pelo número de clusters que você determinou
cluster_and_plot_tsne(sparse, n_clusters_escolhido)

# Você pode experimentar com diferentes valores para perplexity e n_iter para t-SNE
# cluster_and_plot_tsne(sparse, n_clusters_escolhido, perplexity=50, n_iter=500)
```

```python
# prompt: Crie uma função de machine learning para agrupamento usando o algoritmo meanshift e o t-SNE para redução de dimensionalidade, faça no máximo 4 clusters

from sklearn.cluster import MeanShift, estimate_bandwidth

def meanshift_tsne_clustering(X, max_clusters=4, perplexity=30.0, n_iter=300):
    """
    Aplica redução de dimensionalidade com t-SNE e clusterização com Mean Shift.

    Parâmetros:
    - X: matriz de entrada (numpy array ou DataFrame) com n amostras e m atributos.
    - max_clusters: Número máximo de clusters a serem formados pelo Mean Shift (aproximado).
                    O Mean Shift não garante um número exato de clusters, mas tenta
                    encontrar clusters densos. Este parâmetro pode influenciar a largura
                    de banda estimada.
    - perplexity: parâmetro perplexity para t-SNE.
    - n_iter: número de iterações para t-SNE.
    """
    # Redução de dimensionalidade com t-SNE
    print(f"Aplicando t-SNE com perplexity={perplexity}, n_iter={n_iter}...")
    # Use init="random" para entradas esparsas, se X for esparso
    # Se X é denso após algum pré-processamento, init="pca" pode ser mais rápido
    if isinstance(X, (np.ndarray, pd.DataFrame)):
         # Assume que X é denso neste ponto para TSNE
         tsne = TSNE(n_components=2, random_state=42, perplexity=perplexity, n_iter=n_iter, init="pca")
    else: # Assume que X pode ser esparso ou precisa de init="random"
        tsne = TSNE(n_components=2, random_state=42, perplexity=perplexity, n_iter=n_iter, init="random")


    X_tsne = tsne.fit_transform(X)
    print("t-SNE concluído.")

    # Estimar largura de banda (bandwidth) para o Mean Shift
    # A largura de banda é um parâmetro crucial para o Mean Shift.
    # estimate_bandwidth tenta encontrar um valor adequado, mas pode ser sensível.
    # Podemos tentar ajustar o quantile ou usar um valor fixo se necessário.
    # O número de clusters é controlado indiretamente pela largura de banda.
    # Tentar limitar max_clusters aqui é um pouco contra a natureza do Mean Shift,
    # que encontra o número de clusters automaticamente. No entanto, podemos influenciar
    # isso ajustando o quantile ou a largura de banda diretamente.
    # Para forçar um número *aproximado* de clusters, poderíamos estimar a largura
    # de banda e talvez ajustá-la manualmente se o número resultante for muito alto.
    # Uma abordagem alternativa para garantir *exatamente* um número máximo é usar
    # KMeans após o t-SNE, mas a requisição pede Mean Shift.

    # Vamos estimar a largura de banda. Podemos ajustar o quantile se o número de
    # clusters for inconsistentemente alto/baixo. Valores comuns estão entre 0.1 e 0.3.
    # Reduzir o quantile geralmente resulta em mais clusters. Aumentar o quantile
    # geralmente resulta em menos clusters.
    quantile_val = 0.2 # Pode ajustar este valor para controlar o número de clusters
    bandwidth = estimate_bandwidth(X_tsne, quantile=quantile_val, n_samples=min(1000, X_tsne.shape[0])) # Usa uma amostra para ser mais rápido

    print(f"Largura de banda estimada: {bandwidth}")

    # Aplica o Mean Shift nos dados reduzidos com a largura de banda estimada
    print("Aplicando Mean Shift...")
    # Use bin_seeding=True para velocidade, mas pode afetar a precisão
    ms = MeanShift(bandwidth=bandwidth, bin_seeding=True)
    ms.fit(X_tsne)
    labels = ms.labels_
    cluster_centers = ms.cluster_centers_

    n_clusters_ = len(np.unique(labels))
    print(f"Mean Shift concluiu. Número de clusters encontrados: {n_clusters_}")

    # Se o número de clusters for maior que max_clusters, podemos ajustar a largura de banda
    # ou simplesmente notar que Mean Shift encontrou mais agrupamentos densos.
    # Não vamos forçar o número exato aqui, pois Mean Shift é exploratório.
    # A visualização mostrará os clusters encontrados.

    # Plotando os clusters
    plt.figure(figsize=(10, 8))
    scatter = plt.scatter(X_tsne[:, 0], X_tsne[:, 1], c=labels, cmap='viridis', s=10)
    plt.scatter(cluster_centers[:, 0], cluster_centers[:, 1], c='red', s=200, alpha=0.75, marker='X', label='Centros')

    plt.title(f'Clusterização com Mean Shift ({n_clusters_} clusters) e t-SNE')
    plt.xlabel('Componente t-SNE 1')
    plt.ylabel('Componente t-SNE 2')
    plt.legend()
    plt.grid(True)
    plt.colorbar(scatter, label='Cluster')
    plt.show()

# Exemplo de uso com Mean Shift e t-SNE:
# Use a matriz esparsa 'sparse' criada no código anterior
# Certifique-se de que 'sparse' esteja disponível e seja a entrada desejada.
# Se 'sparse' é scipy csr_matrix, talvez precise convertê-la para densa para t-SNE
# ou verificar se a versão de t-SNE suporta matrizes esparsas.
# sklearn.manifold.TSNE suporta matrizes esparsas a partir da versão 0.22.

# Verificando o tipo de 'sparse'
print(type(sparse))

# Se 'sparse' é uma matriz esparsa scipy (csr_matrix), ela pode ser passada diretamente para TSNE
# Se você estiver usando uma versão mais antiga do sklearn ou encontrar problemas,
# pode precisar converter para densa: X_dense = sparse.todense()
# Cuidado: converter para denso pode consumir muita memória se a matriz for grande.

meanshift_tsne_clustering(sparse, max_clusters=4, perplexity=50, n_iter=500)

# Você pode experimentar com diferentes valores para perplexity e n_iter para t-SNE,
# e ajustar o 'quantile' dentro da função para influenciar a largura de banda do Mean Shift.
# meanshift_tsne_clustering(sparse, max_clusters=4, perplexity=30, n_iter=300)

```

```python
# prompt: Crie uma função de machine learning para agrupamento usando o algoritmo  Hierarchical Clustering e o PCA para redução de dimensionalidade, com 4 clusters

from sklearn.cluster import AgglomerativeClustering
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np # Import numpy
import pandas as pd # Import pandas for DataFrame check
from scipy.sparse import csr_matrix # Import csr_matrix for isinstance check


def hierarchical_pca_clustering(X, n_clusters):
    """
    Aplica redução de dimensionalidade com PCA, e agrupamento usando o algoritmo
    Hierarchical Clustering.

    Parâmetros:
    - X: matriz de entrada (numpy array, DataFrame, ou matriz esparsa scipy)
         com n amostras e m atributos.
    - n_clusters: número de clusters desejado.
    """
    print("Aplicando PCA para redução de dimensionalidade...")
    # Aplica PCA. Reduz para um número menor de componentes que capturam a variância.
    # Um número razoável pode ser 0.95 para reter 95% da variância, ou um número fixo
    # que seja significativamente menor que as features originais.
    # Para visualização, muitas vezes reduzimos para 2 componentes.
    # Se n_components é menor que o número original de features, ele reduz.
    # Se n_components for float (entre 0 e 1), seleciona o número de componentes
    # para explicar essa proporção da variância.
    # Se X for esparso, PCA em scikit-learn geralmente funciona melhor com entrada densa.
    # Cuidado com o consumo de memória ao converter para denso.
    # Vamos usar um número fixo de componentes para este exemplo, digamos 2 para fácil visualização.
    # Se seus dados forem muito esparsos e grandes, considere IncrementalPCA.
    # Para matrizes esparsas csr_matrix, .todense() pode falhar por falta de memória.
    # Neste caso, você pode precisar aplicar PCA em chunks ou usar técnicas que lidam
    # diretamente com esparsidade ou PCA incremental.

    # Para este exemplo, vamos assumir que X pode ser convertido para denso para PCA,
    # mas esteja ciente desta limitação com dados muito grandes e esparsos.
    # Se 'sparse' do código anterior é scipy csr_matrix, .todense() converte.
    # Se você estiver usando uma entrada diferente para esta função, ajuste conforme necessário.

    if isinstance(X, (np.ndarray, pd.DataFrame)):
        X_dense = X
    elif isinstance(X, csr_matrix): # Verifica se é uma matriz esparsa scipy csr_matrix
        try:
            # Fix: Convert to numpy array
            X_dense = np.asarray(X.todense())
            print("Matriz esparsa convertida para densa e numpy array para PCA.")
        except MemoryError:
            print("Erro de memória ao converter matriz esparsa para densa. Considere usar IncrementalPCA ou outra abordagem.")
            return # Ou implemente uma alternativa como IncrementalPCA
    else:
        print("Tipo de entrada X não suportado para PCA neste formato.")
        return


    pca = PCA(n_components=2, random_state=42) # Reduz para 2 componentes para visualização
    X_pca = pca.fit_transform(X_dense)
    print(f"PCA concluído. Dados reduzidos para {X_pca.shape[1]} dimensões.")

    # Aplica o algoritmo Hierarchical Clustering
    print(f"Aplicando Hierarchical Clustering com {n_clusters} clusters...")
    # affinity='euclidean' e linkage='ward' são escolhas comuns.
    # 'ward' minimiza a variância dos clusters que estão sendo unidos.
    # metric lida com a métrica de distância.
    # Fix: Replace 'affinity' with 'metric'
    hierarchical_clustering = AgglomerativeClustering(n_clusters=n_clusters, metric='euclidean', linkage='ward')
    labels = hierarchical_clustering.fit_predict(X_pca) # Aplica o clustering nos dados reduzidos pelo PCA
    print("Hierarchical Clustering concluído.")

    # Plotando os clusters nos componentes PCA
    plt.figure(figsize=(10, 8))
    scatter = plt.scatter(X_pca[:, 0], X_pca[:, 1], c=labels, cmap='viridis', s=10)

    plt.title(f'Clusterização com Hierarchical Clustering ({n_clusters} clusters) e PCA')
    plt.xlabel('Componente PCA 1')
    plt.ylabel('Componente PCA 2')
    plt.grid(True)
    plt.colorbar(scatter, label='Cluster')
    plt.show()

# Exemplo de uso da nova função:
# Use a matriz 'sparse' ou 'encoded_df' ou qualquer outra matriz que você queira clusterizar
# Vamos usar 'sparse' do código anterior, que é uma matriz esparsa csr_matrix
# Lembre-se que a conversão para denso pode ser um problema com dados grandes.

# Use a matriz que você quer clusterizar. Ex: sparse
# sparse foi carregada/criada anteriormente como uma csr_matrix
hierarchical_pca_clustering(sparse, n_clusters=4)
```

```python
# prompt: Crie uma função de machine learning para agrupamento usando o algoritmo  Gaussian Mixture Models e o PCA para redução de dimensionalidade, com 4 clusters

from sklearn.mixture import GaussianMixture

def gmm_pca_clustering(X, n_clusters):
    """
    Aplica redução de dimensionalidade com PCA e agrupamento usando o algoritmo
    Gaussian Mixture Models.

    Parâmetros:
    - X: matriz de entrada (numpy array, DataFrame, ou matriz esparsa scipy)
         com n amostras e m atributos.
    - n_clusters: número de clusters desejado.
    """
    print("Aplicando PCA para redução de dimensionalidade...")
    # Aplica PCA. Reduz para um número menor de componentes.
    # Para GMM, é recomendado que o número de componentes PCA seja menor que n_samples.
    # Vamos reduzir para 2 componentes para fácil visualização.
    # Cuidado com a conversão de matrizes esparsas para densas.
    if isinstance(X, (np.ndarray, pd.DataFrame)):
        X_dense = X
    elif isinstance(X, csr_matrix): # Verifica se é uma matriz esparsa scipy csr_matrix
        try:
            # Convert to numpy array
            X_dense = np.asarray(X.todense())
            print("Matriz esparsa convertida para densa e numpy array para PCA.")
        except MemoryError:
            print("Erro de memória ao converter matriz esparsa para densa. Considere usar IncrementalPCA ou outra abordagem.")
            return # Ou implemente uma alternativa como IncrementalPCA
    else:
        print("Tipo de entrada X não suportado para PCA neste formato.")
        return

    # Determine um número de componentes PCA razoável se a entrada for muito grande
    # ou use 2 para visualização conforme a função original de PCA.
    # Para GMM, o número de componentes deve ser <= o número de amostras.
    # Se n_clusters for 4, 2 componentes PCA são suficientes para visualização,
    # mas você pode querer usar mais componentes PCA para o GMM real se a dimensão original for muito alta.
    # Vamos usar 2 para manter a consistência com as funções de plotagem anteriores.
    n_components_pca = 2
    if X_dense.shape[1] < n_components_pca:
        n_components_pca = X_dense.shape[1]
        print(f"Número de componentes PCA ajustado para {n_components_pca} pois a dimensão original é menor.")


    pca = PCA(n_components=n_components_pca, random_state=42) # Reduz para n_components_pca para visualização
    X_pca = pca.fit_transform(X_dense)
    print(f"PCA concluído. Dados reduzidos para {X_pca.shape[1]} dimensões.")

    # Aplica o algoritmo Gaussian Mixture Models
    print(f"Aplicando Gaussian Mixture Models com {n_clusters} componentes (clusters)...")
    # n_components é o número de distribuições gaussianas (clusters).
    # covariance_type pode ser 'full', 'tied', 'diag', 'spherical'. 'full' é o default.
    # n_init: número de inicializações para executar. O melhor resultado é retido.
    gmm = GaussianMixture(n_components=n_clusters, random_state=42, n_init=10)
    gmm.fit(X_pca)
    labels = gmm.predict(X_pca)
    print("Gaussian Mixture Models concluído.")

    # Plotando os clusters nos componentes PCA
    plt.figure(figsize=(10, 8))
    scatter = plt.scatter(X_pca[:, 0], X_pca[:, 1], c=labels, cmap='viridis', s=10, alpha=0.6)

    # O GMM não tem "centros" no mesmo sentido do KMeans, mas podemos plotar as médias das distribuições
    # Note que estes são os centros das distribuições gaussianas ajustadas, não necessariamente
    # a média dos pontos em cada cluster atribuído.
    cluster_means = gmm.means_
    plt.scatter(cluster_means[:, 0], cluster_means[:, 1], c='red', s=200, alpha=0.75, marker='X', label='Médias GMM')


    plt.title(f'Clusterização com Gaussian Mixture Models ({n_clusters} clusters) e PCA')
    plt.xlabel('Componente PCA 1')
    plt.ylabel('Componente PCA 2')
    plt.grid(True)
    plt.colorbar(scatter, label='Cluster')
    plt.legend()
    plt.show()

# Exemplo de uso da nova função:
# Use a matriz 'sparse' ou 'encoded_df' ou qualquer outra matriz que você queira clusterizar
# Vamos usar 'sparse' do código anterior, que é uma matriz esparsa csr_matrix
# Lembre-se que a conversão para denso pode ser um problema com dados grandes.

# Use a matriz que você quer clusterizar. Ex: sparse
# sparse foi carregada/criada anteriormente como uma csr_matrix
gmm_pca_clustering(sparse, n_clusters=4)
```

```python
def recomendar_filmes_por_cluster(filme_titulo: str, clustered_df: pd.DataFrame, num_recomendacoes: int = 5) -> list:
    """
    Recebe o título de um filme e retorna uma lista de recomendações de filmes
    que pertencem ao mesmo cluster.

    Parâmetros:
    - filme_titulo: O título do filme para o qual buscar recomendações.
    - clustered_df: O DataFrame pandas que contém os filmes e a coluna 'cluster'.
    - num_recomendacoes: O número de filmes a serem recomendados (padrão é 5).

    Retorna:
    - Uma lista de strings contendo os títulos dos filmes recomendados,
      ou uma mensagem de erro se o filme não for encontrado ou o cluster não for definido.
    """

    filme_info = clustered_df[clustered_df['title'].str.lower() == filme_titulo.lower()]

    if filme_info.empty:
        return [f"Filme '{filme_titulo}' não encontrado no DataFrame."]

    if 'cluster' not in filme_info.columns or filme_info['cluster'].iloc[0] is None:
         return [f"O cluster para o filme '{filme_titulo}' não está definido."]


    cluster_do_filme = filme_info['cluster'].iloc[0]

    filmes_no_mesmo_cluster = clustered_df[
        (clustered_df['cluster'] == cluster_do_filme) &
        (clustered_df['title'].str.lower() != filme_titulo.lower())
    ]

    if filmes_no_mesmo_cluster.empty:
        return [f"Não foram encontrados outros filmes no cluster {cluster_do_filme} do filme '{filme_titulo}'."]

    num_disponiveis = len(filmes_no_mesmo_cluster)
    num_a_recomendar = min(num_recomendacoes, num_disponiveis)

    if num_a_recomendar == 0:
         return [f"Não há filmes suficientes no cluster {cluster_do_filme} para fornecer recomendações."]


    recomendacoes = filmes_no_mesmo_cluster.sample(n=num_a_recomendar)['title'].tolist()

    return recomendacoes

titulo_filme_exemplo = "Forrest Gump"
recomendacoes = recomendar_filmes_por_cluster(titulo_filme_exemplo, df_clustered, num_recomendacoes=5)

print(f"\nRecomendações para o filme '{titulo_filme_exemplo}':")
for rec in recomendacoes:
    print(f"- {rec}")

titulo_filme_exemplo = "Venom"
recomendacoes = recomendar_filmes_por_cluster(titulo_filme_exemplo, df_clustered, num_recomendacoes=5)

print(f"\nRecomendações para o filme '{titulo_filme_exemplo}':")
for rec in recomendacoes:
    print(f"- {rec}")

titulo_filme_exemplo = "Barbie"
recomendacoes = recomendar_filmes_por_cluster(titulo_filme_exemplo, df_clustered, num_recomendacoes=5)

print(f"\nRecomendações para o filme '{titulo_filme_exemplo}':")
for rec in recomendacoes:
    print(f"- {rec}")
```
